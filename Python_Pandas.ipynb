```python
# 84.	Detect BMI outliers using IQR
# 📝 Note:
# o	Lower bound = Q1 − 1.5×IQR
# o	Upper bound = Q3 + 1.5×IQR
# 1. Calculate Q1 (25th percentile) and Q3 (75th percentile)
q1 = df1['BMI'].quantile(0.25)
q3 = df1['BMI'].quantile(0.75)

# 2. Calculate the Interquartile Range (IQR)
iqr = q3 - q1

# 3. Determine upper and lower bounds
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

# 4. Filter out the outliers
bmi_outliers = df1[(df1['BMI'] < lower_bound) | (df1['BMI'] > upper_bound)]

print(f"Lower Bound: {lower_bound:.2f}, Upper Bound: {upper_bound:.2f}")
print(f"Total outliers detected: {len(bmi_outliers)}")
print(bmi_outliers[['Patient_ID', 'BMI']])
```
