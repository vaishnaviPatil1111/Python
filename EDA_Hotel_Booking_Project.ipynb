#### Duplicate Values:

````python
#Finding duplicate values using count function
df.duplicated().value_counts()
````

````python
# Visualizing the duplicate values
plt.figure(figsize = (5,5))
#assigning the fig size to this countplot
#For better Well-structured and code formatted we will use (11,9) fig size to evry chart

sns.countplot(x=df.duplicated(),color='Orange')
plt.title('Visualisation of duplicated value')
plt.ylabel('Duplicate values')
plt.show()
````

#### Missing Values/Null Values:

````python
# Missing Values/Null Values Count
df.isna() == True
#we have check the null values in DataFrame
````

````python
#We have to fill all null values with nan
df.fillna(np.nan, inplace = True)
df
````

````python
# Visualizing the missing values
a = df.isna().sum()
missing_values = a.sort_values(ascending = False)[:5]
missing_values
````

````python
plt.figure(figsize = (7,5))
missing_values.plot(kind='bar')
plt.title('Visualisation of missing value', color='Red')
plt.ylabel('missing value', color = 'Red')
plt.xlabel('Rows', color='Red')
plt.xticks(rotation=0)
plt.show()
````
